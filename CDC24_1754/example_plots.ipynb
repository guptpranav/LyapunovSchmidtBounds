```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

plt.rcParams.update({'font.size': 21.75})
```

```python
p = 0.9

# Define the system of ODEs
def dx1_dt(x1, x2):
    return -x1 + np.tanh(p * x2)

def dx2_dt(x1, x2):
    return -x2 + np.tanh(p * x1)

# Define the grid of points
x1 = np.linspace(-1.1, 1.1, 20)
x2 = np.linspace(-1.1, 1.1, 20)
X1, X2 = np.meshgrid(x1, x2)

# Calculate the derivatives at each point
dX1_dt = dx1_dt(X1, X2)
dX2_dt = dx2_dt(X1, X2)

# Plot the vector field
#plt.quiver(X1, X2, dX1_dt, dX2_dt)
plt.streamplot(X1, X2, dX1_dt, dX2_dt, color='black')

# plot red circle at the origin 
plt.plot(0, 0, 'ro',markersize=10)
plt.plot(0, 0, 'ro',markersize=10)

# find zeros of tanh(p*tanh(p*x)) = 0
x = fsolve(lambda x: np.tanh(p*np.tanh(p*x))-x, 0.5)
plt.plot(x, x, 'ro',markersize=10)
plt.plot(-x, -x, 'ro',markersize=10)

plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.title(r'$\lambda = $ '+str(p), usetex=True)
plt.savefig('images/0_9.svg', dpi=600, format='svg')
plt.show()
```
